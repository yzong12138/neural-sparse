```python
import os
import datasets
ds=[]
# for path_ in ["data/pt-kd","data/beir-kd"]:
for path_ in ["data/beir-kd-v2"]:
    for dataset in os.listdir(path_):
        dataset_path = os.path.join(path_, dataset)
        ds.append(datasets.load_from_disk(dataset_path))
        
# ds.append(datasets.load_from_disk("data/msmarco_ft"))
# ds.append(datasets.load_from_disk("data/msmarco_ft2"))
```

```python
import pickle

with open("data/query_to_id.bin","rb") as f:
    query_to_id = pickle.load(f)
    
with open("data/doc_to_id.bin","rb") as f:
    doc_to_id = pickle.load(f)
```

```python
len(query_to_id), len(doc_to_id)
```

```python
from tqdm import tqdm

print(len(query_to_id))

num = len(query_to_id)
for d in ds[:]:
    for x in tqdm(d):
        # texts =[x["query"]] + x["docs"]
        text = x["query"]
        if text not in query_to_id:
            query_to_id[text] = num
            num += 1
            
print(len(query_to_id))
```

```python
import pickle

with open("data/query_to_id.bin","wb") as f:
    pickle.dump(query_to_id, f)
    
```

```python
from tqdm import tqdm

print(len(doc_to_id))

num = len(doc_to_id)
for d in ds[:]:
    for x in tqdm(d):
        texts = x["docs"]
        for text in texts:
            if text not in doc_to_id:
                doc_to_id[text] = num
                num += 1
                
print(len(doc_to_id))
```

```python
import pickle

print(len(doc_to_id))

with open("data/doc_to_id.bin","wb") as f:
    pickle.dump(doc_to_id, f)
```
